## Slide 11 · 变量与基本数据类型

````python
# ── 四种基本类型 ──────────────────────────────────────────────────────────
# Slide 11 · 强调与R的对照（右侧注释）
age = 61              # int     —— R: numeric (integer)
egfr_expr = 8.52      # float   —— R: numeric
gene = "EGFR"         # str     —— R: character
is_mutated = True     # bool    —— R: TRUE

print(type(age), type(egfr_expr), type(gene), type(is_mutated))

# ⚡ 互动：把 61 改成 "61岁" 再运行，类型会变吗？
type(gene)            # R: class(gene)
````

````python
# ── 动态类型与类型转换 ─────────────────────────────────────────────────────
x = 61            # 现在是 int
x = "sixty-one"   # 现在是 str —— 类型跟着值走！
print(type(x))

# 转换三件套
print(int("61"), str(8.52), float("3.14"))

# ⏸ 故意留错：字符串 + 整数
"61" + 1
````

> 💬 **讨论点**：R 里 `"61" + 1` 是报错，Python 里是直接 `TypeError`——两种语言都不允许，但 Python 的报错信息更清晰。

## Slide 12 · 四大核心数据结构

````python
# ── list / tuple / dict / set 对照表 ────────────────────────────────────
# Slide 12
xs = [3, 5, 7]                  # 列表 list   —— 有序可变   —— R: 向量(可混类型)
t  = (3, "EGFR")                # 元组 tuple  —— 有序不可变 —— R: 无对应
d  = {"gene": "EGFR", "expr": 8.5}  # 字典 dict  —— 键值对   —— R: 命名列表
s  = {3, 5, 5, 7}               # 集合 set    —— 无序去重   —— R: unique()

print("list :", xs, type(xs).__name__)
print("tuple:", t, type(t).__name__)
print("dict :", d, type(d).__name__)
print("set  :", s, type(s).__name__, " <- 5只出现一次！")
````

## Slide 13 · 列表：0-based 索引（第一坑）

````python
# ── ⚡ 0-based 索引演示 ──────────────────────────────────────────────────
# Slide 13 · 口诀：Python从0开始数；切片[a:b]拿a不拿b
expr = [8.5, 2.1, 9.7, 4.3]

print("expr[0]  =", expr[0],    "  <- 第1个元素！ R: expr[1]")
print("expr[1:3]=", expr[1:3],  "  <- 含头不含尾 R: expr[2:3]")
print("expr[-1] =", expr[-1],   "  <- 倒数第1个")

expr.append(6.0)   # R: c(expr, 6.0)
print("追加后    :", expr, " len =", len(expr))
````

````python
# ── ⏸ 故意留错：索引越界 ──────────────────────────────────────────────────
# R: expr[5] 返回 NA；Python: 直接报 IndexError
expr = [8.5, 2.1, 9.7, 4.3]
expr[5]
````

## Slide 16 · 函数定义：GC含量案例

这个函数在**12月实验课**和**第15讲（序列绘图）**还会用到——函数是复用单元。

````python
# ── 函数：def + docstring + 默认参数 ────────────────────────────────────
# Slide 16 · R: gc <- function(seq) {...}
def gc_content(seq, digits=1):
    """计算DNA序列的GC含量(%)"""
    seq = seq.upper()
    gc = seq.count("G") + seq.count("C")
    return round(gc / len(seq) * 100, digits)

print(gc_content("ATGGCGTAA"))        # 位置参数
print(gc_content(seq="ATGGCGTAA"))    # 关键字参数
print(gc_content("ATGGCGTAA", 3))     # 覆盖默认精度 -> 44.444

# help(gc_content)   # 演示：docstring 就在这里生效
````

## Slide 17 · 列表推导式：对应 R 的向量化写法

````python
# ── 列表推导式 vs R sapply ──────────────────────────────────────────────
# Slide 17 · R: sapply(c("ATGC","GGCC"), gc)
result = [gc_content(s) for s in ["ATGC", "GGCC"]]
print(result)   # [50.0, 100.0]

# ⚡ 带条件的推导式（R: 只对满足条件的元素计算）
result2 = [gc_content(s) for s in ["ATGC", "GGCC", "AAAA"] if "G" in s]
print(result2)  # AAAA 被过滤掉

# 铺垫：NumPy 才是真正的向量化 —— 下一讲介绍（快100倍）
````

## Slide 29 · 布尔索引：条件筛选（重点+难点）

````python
# ── 单条件 ──────────────────────────────────────────────────────────────
# R: df[df$age > 60, ] 或 filter(df, age > 60)
df[df["age"] > 60]
````

````python
# ── 多条件：括号！ & 和 |（不是 && ||）──────────────────────────────────
elderly_smokers = df[(df["age"] > 60) & (df["smoking"] == "yes")]
print(f"年龄>60 且 吸烟: {len(elderly_smokers)} 人")
elderly_smokers[["patient_id", "age", "smoking", "stage"]]
````

````python
# ── ⚠️ 故意写错：忘括号 —— 比报错更危险的是不报错！ ────────────────────
# & 的优先级高于比较：实际执行的是 df["age"] > (60 & (df["smoking"] == "yes"))
# 60 & 布尔列 → 全变 0，于是 df["age"] > 0 恒真 —— 整表误筛且不报错
wrong = df[df["age"] > 60 & (df["smoking"] == "yes")]
print(f"忘括号的结果: {len(wrong)} 人   <- 全部30人都回来了，明显不对！")
print("字符串条件忘括号直接 TypeError；数值/布尔列则静默误筛 —— 更要警惕")
print("\n正确写法（每个条件必须加括号）:")
right = df[(df["age"] > 60) & (df["smoking"] == "yes")]
print(f"正确的结果  : {len(right)} 人")
````

````python
# ── 字符串筛选 + query() 写法 ────────────────────────────────────────────
print("以LC0开头的编号:")
print(df[df["patient_id"].str.startswith("LC0")].shape[0], "人")

# query() 像SQL，可读性好（结果与上面布尔索引完全等价）
df.query("age > 60 and smoking == 'yes'").head()
````

## Slide 30 · 缺失值处理

> 💬 **思政点**：删除/填补前，先记录行数并说明理由——以透明、可追溯的方式报告数据排除标准。

````python
# ── 缺失值三连 ───────────────────────────────────────────────────────────
# R: colSums(is.na(df))
print("每列缺失数:")
print(df.isnull().sum())

# 哪些行有缺失？
print("\n含缺失的患者:")
print(df[df.isnull().any(axis=1)][["patient_id", "egfr_expr", "pd_l1_tps", "survival_months"]])
````

````python
# ── 删除 vs 填补 ─────────────────────────────────────────────────────────
print("原始行数        :", len(df))
print("dropna后        :", len(df.dropna(subset=["egfr_expr"])), "<- R: df[!is.na(df$egfr_expr),]")

# 中位数填补（作业Q3就是这道题！答案: 6.67）
median_val = df["egfr_expr"].median()
filled = df["egfr_expr"].fillna(median_val)
print(f"\negfr_expr 中位数 = {median_val}")
print(f"填补前均值 = {df['egfr_expr'].mean():.2f}")
print(f"填补后均值 = {filled.mean():.2f}   <- 作业Q3答案")
````

## Slide 31 · 增、删、改列

````python
# ── mutate / select / rename 在 Pandas 中的对应写法 ─────────────────────
# Slide 31
work = df.copy()   # 演示用副本，保护原数据

# 新增列（R: mutate + cut）
work["age_group"] = pd.cut(work["age"],
                           bins=[0, 50, 70, 120],
                           labels=["<50", "50-70", ">70"])
print("age_group 分布:")
print(work["age_group"].value_counts().sort_index())

# 删除列（R: select(-status)）
work = work.drop(columns=["status"])
print("\n删除status后的列:", list(work.columns))

# 重命名（R: rename(pid = patient_id)）
work = work.rename(columns={"patient_id": "pid"})
print("重命名后的列 :", list(work.columns))
````
